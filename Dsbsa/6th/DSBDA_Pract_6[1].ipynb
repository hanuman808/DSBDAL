```python
import numpy as nm  
```

```python
import pandas as pd  
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
dataset = pd.read_csv("Social_Network_Ads.csv")
x = dataset.iloc[:, [2, 3]].values  
y = dataset.iloc[:, 4].values  
```

```python
dataset.head(5)
```

```python
from sklearn.model_selection import train_test_split  
x_train, x_test, y_train, y_test = train_test_split(x, 
            y, test_size = 0.25, random_state = 0)  
```

```python
from sklearn.preprocessing import StandardScaler  
sc = StandardScaler()
```

```python
x_train = sc.fit_transform(x_train)  
x_test = sc.transform(x_test)  
```

```python
from sklearn.naive_bayes import GaussianNB  
classifier = GaussianNB()  
classifier.fit(x_train, y_train)  
```

```python
GaussianNB(priors=None, var_smoothing=1e-09)
```

```python
y_pred = classifier.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
```

```python
from sklearn.preprocessing import StandardScaler
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import matplotlib.pyplot as mtp


sc = StandardScaler()
x_train = sc.fit_transform(x_train)  
x_test = sc.transform(x_test)


x_set, y_set = x_train, y_train


X1, X2 = np.meshgrid(np.arange(start=x_set[:, 0].min() - 1, 
                     stop=x_set[:, 0].max() + 1, step=0.01),
                     np.arange(start=x_set[:, 1].min() - 1, 
                     stop=x_set[:, 1].max() + 1, step=0.01))


mtp.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), 
                    X2.ravel()]).T).reshape(X1.shape),
                    alpha=0.75, 
                    cmap=LinearSegmentedColormap.from_list
                    ("custom_cmap", ['blue', 'orange']))


mtp.xlim(X1.min(), X1.max())
mtp.ylim(X2.min(), X2.max())


for i, j in enumerate(np.unique(y_set)):
    
    mtp.scatter(x_set[y_set == j, 0], x_set[y_set == j, 1],
                c=LinearSegmentedColormap.from_list
                ("custom_cmap", ['blue', 'orange'])(i), label=j)


mtp.title('Naive Bayes (Training set)')
mtp.xlabel('Age')
mtp.ylabel('Estimated Salary')


mtp.legend()


mtp.show()
```

```python
from sklearn.preprocessing import StandardScaler
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import matplotlib.pyplot as mtp


sc = StandardScaler()
x_train = sc.fit_transform(x_train)  
x_test = sc.transform(x_test)

# Set the data for plotting
x_set, y_set = x_train, y_train


X1, X2 = np.meshgrid(np.arange(start=x_set[:, 0].min() - 1, 
                     stop=x_set[:, 0].max() + 1, step=0.01),
                     np.arange(start=x_set[:, 1].min() - 1, 
                     stop=x_set[:, 1].max() + 1, step=0.01))


mtp.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), 
                     X2.ravel()]).T).reshape(X1.shape),
                     alpha=0.75, 
                     cmap=LinearSegmentedColormap.from_list
                     ("custom_cmap", ['blue', 'orange']))


mtp.xlim(X1.min(), X1.max())
mtp.ylim(X2.min(), X2.max())

colors = ['green', 'pink']  

for i, j in enumerate(np.unique(y_set)):
    mtp.scatter(x_set[y_set == j, 0], x_set[y_set == j, 1],
                c=colors[i], label=f'Class {j}')  


mtp.title('Naive Bayes (Training set)')
mtp.xlabel('Age')
mtp.ylabel('Estimated Salary')

# Display legend
mtp.legend()

# Show the plot
mtp.show()
```
